```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv("air_quality_dataset.csv")
print(df.head())
```

```python
print(df.info())
```

```python
df.columns[df.any()]
```

```python
df.columns[df.isnull().any()]
```

```python
AQI_mean=int(df["AQI"].mean())
print(AQI_mean)
```

```python
PM2_5_mean=df["PM2.5"].mean()
print(round(PM2_5_mean,2))
```

```python
PM10_mean=df["PM10"].mean()
print(round(PM10_mean,2))
```

```python
Ozone_mean=df["Ozone"].mean()
print(round(Ozone_mean,2))
```

```python
NO2_mean=df["NO2"].mean()
print(round(NO2_mean,2))
```

```python
CO_mean=df["CO"].mean()
print(round(CO_mean,3))
```

```python
SO2_mean=df["SO2"].mean()
print(round(SO2_mean,2))
```

```python
Temperature_mean=df["Temperature"].mean()
print(round(Temperature_mean,1))
```

```python
Humidity_mean=df["Humidity"].mean()
print(round(Humidity_mean,1))
```

```python
new_df=df.fillna({
    "AQI":AQI_mean,
    "PM2.5":PM2_5_mean,
    "PM10":PM10_mean,
    "Ozone":Ozone_mean,
    "NO2":NO2_mean,
    "CO":CO_mean,
    "SO2":SO2_mean,
    "Temperature":Temperature_mean,
    "Humidity":Humidity_mean
})
print(new_df)
```

```python
print(new_df.info())
```

```python
new_df.columns[new_df.isnull().any()]
```

```python
new_df.to_csv("clean_data.csv", index=False)
print("Clean data saved successfully!")
```

```python
city_aqi = df.groupby("City")["AQI"].mean().sort_values(ascending=False)

sns.barplot(x=city_aqi.values, y=city_aqi.index)
plt.xlabel("Average AQI")
plt.ylabel("City")
plt.show()
```
